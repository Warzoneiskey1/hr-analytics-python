# HR Analytics – Jeffrey Godwin

Analyzed 1,200+ employee records for attrition, salary trends, and diversity using Python Pandas.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data (download sample HR CSV from Kaggle or use this example)
df = pd.read_csv('hr_data.csv')
print('Rows:', len(df))
df.head()
```

```python
# Cleaning
df.drop_duplicates(inplace=True)
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})
print('Clean rows:', len(df))
print('Attrition rate:', df['Attrition'].mean() * 100, '%')
```

```python
# Insights
print('Average salary by department:')
print(df.groupby('Department')['Salary'].mean())

plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Department', y='Salary')
plt.title('Average Salary by Department')
plt.xticks(rotation=45)
plt.show()
```
