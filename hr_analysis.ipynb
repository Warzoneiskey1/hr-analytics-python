# HR Analytics – Jeffrey Godwin

Analyzed 1,470+ employee records for attrition, salary trends, and performance using Python Pandas.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data directly from the new free dataset link (no download needed)
df = pd.read_csv('https://raw.githubusercontent.com/ParthRaj22/HR-Analytics/master/HR.csv')
print('Rows:', len(df))
df.head()
```

```python
# Cleaning
df.drop_duplicates(inplace=True)
df['attrition'] = df['attrition'].map({'Yes': 1, 'No': 0})
print('Clean rows:', len(df))
print('Attrition rate:', df['attrition'].mean() * 100, '%')
```

```python
# Insights
print('Average salary by department:')
print(df.groupby('department')['monthly_income'].mean())

plt.figure(figsize=(10,6))
sns.barplot(data=df, x='department', y='monthly_income')
plt.title('Average Monthly Income by Department')
plt.xticks(rotation=45)
plt.show()
```
